# Which month had the highest revenue in 2024?

```sql
SELECT
    _pickup_month AS month,
    total_trips,
    total_revenue_usd,
    avg_fare_usd,
    avg_tip_pct
FROM nyc_taxi_datalake.gold.monthly_trip_metrics
WHERE _pickup_year = 2024
ORDER BY total_revenue_usd DESC
```

#What is the busiest hour of the entire week?


```sql
SELECT
    CASE pickup_dow
        WHEN 1 THEN 'Sunday'
        WHEN 2 THEN 'Monday'
        WHEN 3 THEN 'Tuesday'
        WHEN 4 THEN 'Wednesday'
        WHEN 5 THEN 'Thursday'
        WHEN 6 THEN 'Friday'
        WHEN 7 THEN 'Saturday'
    END AS day_of_week,
    pickup_hour,
    total_trips,
    avg_fare_usd
    FROM nyc_taxi_datalake.gold.hourly_demand_patterns
    WHERE _pickup_year = 2024
    ORDER BY total_trips DESC
    LIMIT 10 
```

#Which top 10 zones generate the most revenue?

```sql
SELECT
    pickup_zone,
    pickup_borough,
    total_pickups,
    total_revenue_usd,
    avg_revenue_per_trip,
    avg_tip_pct
    FROM nyc_taxi_datalake.gold.zone_performance
    WHERE _pickup_year = 2024
    ORDER BY total_revenue_usd DESC
    LIMIT 10
```

#What is the average tip % by borough?

```sql
SELECT
    pickup_borough,
    ROUND(AVG(avg_tip_pct), 2) AS avg_tip_pct,
    SUM(total_pickups) AS total_pickups,
    ROUND(AVG(avg_revenue_per_trip), 2) AS avg_revenue_per_trip
FROM nyc_taxi_datalake.gold.zone_performance
WHERE _pickup_year = 2024
  AND pickup_borough IS NOT NULL
GROUP BY pickup_borough
ORDER BY avg_tip_pct DESC
```
